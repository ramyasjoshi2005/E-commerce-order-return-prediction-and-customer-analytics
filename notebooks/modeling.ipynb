# Modeling — temporal split, comparison, threshold

Metrics below are loaded from `artifacts/` produced by `src/train.py`. They are not hard-coded in source.

```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path('..') if not Path('artifacts/test_metrics.json').exists() else Path('.')
val = pd.read_csv(ROOT / 'artifacts' / 'validation_comparison.csv')
sel = json.loads((ROOT / 'artifacts' / 'model_selection.json').read_text())
test = json.loads((ROOT / 'artifacts' / 'test_metrics.json').read_text())
split = json.loads((ROOT / 'artifacts' / 'split_info.json').read_text())
hyp = json.loads((ROOT / 'artifacts' / 'hypothesis_test.json').read_text())
print(split)
display(val[['model','roc_auc','pr_auc','precision','recall','f1','accuracy','brier_score']])
print('Selected', sel['selected_model'], 'threshold', sel['locked_threshold'])
print('Test', {k: test[k] for k in ['roc_auc','pr_auc','precision','recall','f1','accuracy','brier_score']})
print('z-test p', hyp['p_value'], 'z', hyp['z_statistic'])
```

## Design choices

- **Selection metric:** validation PR-AUC (imbalanced ranking).
- **Threshold:** max F1 on validation, locked before test.
- **CV:** TimeSeriesSplit on training years only.
- Logistic regression won this dataset; XGBoost was not assumed to win.
