# EDA — E-Commerce Return Prediction

This notebook reads **computed** artifacts. Re-run `python -m src.run_pipeline` if files are missing.

Every plot answers a return-rate question. None of these results are causal.

```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path('..') if Path('artifacts').exists() is False else Path('.')
if not (ROOT / 'artifacts' / 'eda.json').exists():
    ROOT = Path('..')
eda = json.loads((ROOT / 'artifacts' / 'eda.json').read_text())
quality = json.loads((ROOT / 'artifacts' / 'data_quality.json').read_text())
print(quality['dataset_name'])
print('orders', quality['n_orders'], 'customers', quality['n_customers'])
print('dates', quality['date_min'], quality['date_max'])
print('return rate', round(quality['return_rate'], 4))
print('duplicate order_ids dropped', quality['duplicate_order_ids_dropped'])
print('India in modeling data', quality['india_in_modeling_data'])
```

## Data quality

- No missing values on modeling columns after cleaning.
- 504 duplicate `order_id` rows were dropped (they leaked the current label into history).
- Labeled table is United States only.

```python
def show(name):
    df = pd.DataFrame(eda[name])
    df['return_rate_pct'] = (df['return_rate'] * 100).round(2)
    display(df)

print('Overall', eda['overall'])
print('\nBy region')
show('return_rate_by_region')
print('\nBy discount')
show('return_rate_by_discount')
print('\nBy order value')
show('return_rate_by_order_value')
print('\nBy prior-return history')
show('return_rate_by_prior_returns')
```
